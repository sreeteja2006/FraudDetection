```python
import pandas as pd
```

```python
df = pd.read_csv("../data/train_transaction.csv")
print(df.shape)
print(df.columns.tolist())
```

```python
print(df['isFraud'].value_counts())
print(df['isFraud'].value_counts(normalize=True) * 100)
```

```python
missing = df.isnull().sum().sort_values(ascending=False)
print(missing[missing > 0].head(20))
```

```python
print(df[['TransactionAmt', 'ProductCD', 'card4', 'card6', 'P_emaildomain']].head(10))
```

```python
import matplotlib.pyplot as plt

df[df['isFraud']==0]['TransactionAmt'].apply(lambda x: min(x, 500)).hist(bins=50, alpha=0.5, label='Not Fraud')
df[df['isFraud']==1]['TransactionAmt'].apply(lambda x: min(x, 500)).hist(bins=50, alpha=0.5, label='Fraud')
plt.legend()
plt.xlabel('Transaction Amount (capped at 500)')
plt.title('Transaction Amount: Fraud vs Not Fraud')
plt.show()
```
